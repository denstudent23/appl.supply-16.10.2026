```python
import io
import matplotlib.pyplot as plt
import requests
from PIL import Image

# URL картинки
url = "https://encrypted-tbnO.gstatic.com/images?q=tbn:ANd9GcRN-jw7fYJfco 2SdZC2nLBRrOCzSBvhJO7Df08VzFv A&s=10"

try:
  # 1. Получение бинарных данных изображения с помощью requests
  response = requests.get(url)
  response.raise_for_status()

  # 2. Обертывание бинарных данных в поток io.BytesIO
  image_bytes = io.BytesIO(response.content)

  # 3. Открытие изображения из потока байт в памяти с помощью PIL
  img = Image.open(image_bytes)

  # 4. Визуализация изображения с помощью matplotlib
  plt.figure(figsize=(6, 6))
  plt.imshow(img)
  plt.axis("off")  # Скрытие сетки и осей
  plt.title("Loaded via io.BytesIO")
  plt.show()

except Exception as e:
  print(f"Произошла ошибка: {e}")
```

```python
import csv
import sqlite3
from faker import Faker

# 1. Faker를 이용해 가상의 CSV 파일 생성하기
# 한국어 설정으로 가상 데이터를 만듭니다.
fake = Faker('ko_KR')
csv_filename = 'users.csv'

print('1. 📂 가상 CSV 파일 생성 중...')
with open(csv_filename, 'w', encoding='utf-8', newline='') as f:
  writer = csv.writer(f)
  # 헤더(열 이름) 작성
  writer.writerow(['name', 'email', 'city'])

  # 5명의 가상 사용자 데이터 작성
  for _ in range(5):
    writer.writerow([fake.name(), fake.email(), fake.city()])

print(f"'{csv_filename}' 파일이 생성되었습니다!\n")


# 2. SQLite 데이터베이스 생성 및 CSV 데이터 저장
db_filename = 'my_database.db'

print('2. 💾 SQLite DB에 CSV 데이터 저장 중...')
# sqlite3.connect를 with 문과 함께 쓰면 자동으로 커밋(commit)되고 연결이 닫힙니다.
with sqlite3.connect(db_filename) as conn:
  cursor = conn.cursor()

  # 테이블 생성 (이미 있다면 무시)
  cursor.execute('''
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT,
            email TEXT,
            city TEXT
        )
    ''')

  # CSV 파일을 열어서 DB에 한 줄씩 삽입
  with open(csv_filename, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)  # 첫 번째 줄(헤더)은 건너뜁니다.
    for row in reader:
      cursor.execute(
          'INSERT INTO users (name, email, city) VALUES (?, ?, ?)', row
      )

print('데이터베이스 저장 완료!\n')


# 3. CRUD(생성, 조회, 수정, 삭제) 실습하기
def practice_crud():
  # 데이터베이스 연결
  with sqlite3.connect(db_filename) as conn:
    cursor = conn.cursor()

    # --- [C]reate: 새로운 데이터 추가하기 ---
    print('--- [C]reate: 새로운 사용자 추가 ---')
    cursor.execute(
        'INSERT INTO users (name, email, city) VALUES (?, ?, ?)',
        ('김철수', 'chulsoo@example.com', '인천'),
    )
    print('👉 "김철수" 데이터가 추가되었습니다.\n')

    # --- [R]ead: 데이터 조회하기 ---
    print('--- [R]ead: 전체 사용자 목록 조회 ---')
    cursor.execute('SELECT * FROM users')
    rows = cursor.fetchall()  # 모든 행 가져오기
    for row in rows:
      print(row)
    print()

    # --- [U]pdate: 데이터 수정하기 ---
    print('--- [U]pdate: 김철수의 거주지 수정 ---')
    cursor.execute(
        'UPDATE users SET city = ? WHERE name = ?', ('제주', '김철수')
    )
    print('👉 "김철수"의 거주지가 "제주"로 변경되었습니다.\n')

    # 수정 결과 확인용 Read
    print('--- [R]ead (확인용): 김철수 검색 ---')
    cursor.execute('SELECT * FROM users WHERE name = ?', ('김철수',))
    print(cursor.fetchone())  # 한 줄만 가져오기
    print()

    # --- [D]elete: 데이터 삭제하기 ---
    print('--- [D]elete: 김철수 데이터 삭제 ---')
    cursor.execute('DELETE FROM users WHERE name = ?', ('김철수',))
    print('👉 "김철수" 데이터가 삭제되었습니다.\n')


# CRUD 함수 실행
practice_crud()
```

```python
import csv
import sqlite3
from faker import Faker

# 한국어 설정으로 Faker 객체 생성
fake = Faker('ko_KR')
csv_filename = 'users.csv'
db_filename = 'app.db'

print('=== 1. Faker로 예제 CSV 파일 만들기 ===')
with open(csv_filename, 'w', encoding='utf-8', newline='') as f:
  writer = csv.writer(f)
  # 헤더 작성
  writer.writerow(['name', 'email', 'city'])

  # 5명의 가상 데이터 생성 후 입력
  for _ in range(5):
    writer.writerow([fake.name(), fake.email(), fake.city()])

print(f"'{csv_filename}' 파일 생성 완료!\n")


print('=== 2. CSV 데이터를 SQLite DB에 저장하기 ===')
# with sqlite3.connect(...)를 쓰면 작업 완료 후 자동 commit 및 연결 종료 처리
with sqlite3.connect(db_filename) as conn:
  cursor = conn.cursor()

  # 테이블 생성
  cursor.execute('''
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT,
            email TEXT,
            city TEXT
        )
    ''')

  # CSV 읽어서 DB에 삽입
  with open(csv_filename, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)  # 첫 줄(헤더) 스킵
    for row in reader:
      cursor.execute(
          'INSERT INTO users (name, email, city) VALUES (?, ?, ?)', row
      )

print('데이터베이스 저장 완료!\n')


print('=== 3. CRUD(생성, 조회, 수정, 삭제) 실습하기 ===')
with sqlite3.connect(db_filename) as conn:
  cursor = conn.cursor()

  # [C]reate: 데이터 추가
  print('--- [C]reate ---')
  cursor.execute(
      'INSERT INTO users (name, email, city) VALUES (?, ?, ?)',
      ('홍길동', 'hong@example.com', '서울'),
  )
  print('👉 "홍길동" 데이터 추가 완료\n')

  # [R]ead: 데이터 조회
  print('--- [R]ead (전체 목록) ---')
  cursor.execute('SELECT * FROM users')
  for row in cursor.fetchall():
    print(row)
  print()

  # [U]pdate: 데이터 수정
  print('--- [U]pdate ---')
  cursor.execute('UPDATE users SET city = ? WHERE name = ?', ('부산', '홍길동'))
  print('👉 "홍길동"의 거주지를 "부산"으로 수정\n')

  # 수정 확인 조회
  print('--- [R]ead (홍길동 검색) ---')
  cursor.execute('SELECT * FROM users WHERE name = ?', ('홍길동',))
  print(cursor.fetchone())
  print()

  # [D]elete: 데이터 삭제
  print('--- [D]elete ---')
  cursor.execute('DELETE FROM users WHERE name = ?', ('홍길동',))
  print('👉 "홍길동" 데이터 삭제 완료\n')

print('=== 모든 실습이 완료되었습니다! ===')
```

```python
import csv
import sqlite3
from faker import Faker

# 설정 변수
fake = Faker('ko_KR')
csv_file = 'sample_users.csv'
db_file = 'sample.db'

print('=== 1. Faker로 예제 CSV 파일 생성 ===')
with open(csv_file, 'w', encoding='utf-8', newline='') as f:
  writer = csv.writer(f)
  writer.writerow(['name', 'email', 'job'])  # 헤더

  # 5명의 가상 데이터 작성
  for _ in range(5):
    writer.writerow([fake.name(), fake.email(), fake.job()])

print(f"'{csv_file}' 파일이 성공적으로 생성되었습니다.\n")


print('=== 2. CSV 데이터를 SQLite DB로 저장 ===')
# with 문을 사용하면 별도의 commit이나 close 없이 자동 처리됩니다.
with sqlite3.connect(db_file) as conn:
  cursor = conn.cursor()

  # 테이블 생성
  cursor.execute('''
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT,
            email TEXT,
            job TEXT
        )
    ''')

  # CSV 읽어서 DB에 입력
  with open(csv_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)  # 헤더 건너뛰기
    for row in reader:
      cursor.execute('INSERT INTO users (name, email, job) VALUES (?, ?, ?)', row)

print('데이터베이스 저장 완료!\n')


print('=== 3. CRUD 실습 (with 구문 활용) ===')
with sqlite3.connect(db_file) as conn:
  cursor = conn.cursor()

  # [C]reate: 데이터 추가
  print('--- [C]reate: 새 사용자 추가 ---')
  cursor.execute(
      'INSERT INTO users (name, email, job) VALUES (?, ?, ?)',
      ('박지성', 'js.park@example.com', '축구선수'),
  )
  print('👉 "박지성" 데이터 추가 완료\n')

  # [R]ead: 전체 데이터 조회
  print('--- [R]ead: 전체 사용자 목록 ---')
  cursor.execute('SELECT * FROM users')
  for row in cursor.fetchall():
    print(row)
  print()

  # [U]pdate: 데이터 수정
  print('--- [U]pdate: 직업 수정 ---')
  cursor.execute(
      'UPDATE users SET job = ? WHERE name = ?', ('지도자', '박지성')
  )
  print('👉 "박지성"의 직업을 "지도자"로 수정 완료\n')

  # [R]ead (확인): 수정된 데이터 단건 조회
  print('--- [R]ead (확인): 박지성 검색 ---')
  cursor.execute('SELECT * FROM users WHERE name = ?', ('박지성',))
  print(cursor.fetchone())
  print()

  # [D]elete: 데이터 삭제
  print('--- [D]elete: 데이터 삭제 ---')
  cursor.execute('DELETE FROM users WHERE name = ?', ('박지성',))
  print('👉 "박지성" 데이터 삭제 완료\n')

print('=== 모든 실습이 깔끔하게 완료되었습니다! ===')
```
